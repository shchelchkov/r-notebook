```python
import http.client

conn = http.client.HTTPConnection("r-function-orderbook:8203")
payload = ""
conn.request("GET", "/api/functions/values/obi_state/BTCUSDT", payload)

res = conn.getresponse()
json_data = res.read()

print(json_data.decode("utf-8"))
```

```python
import json
import pandas as pd

data = json.loads(json_data)

df = pd.DataFrame(data["data"])
```

```python
def values(setting="obi_state",
           code="BTCUSDT"):
    base_url = f"http://r-function-orderbook:8203/api/functions/values/{setting}/{code}"
    response = requests.get(base_url)

    if response.status_code == 200:
        return response.json()
    else:
        print(f"Error: {response.status_code}")
        return None
```

```python
data=values("obi_state", "BTCUSDT")
df = pd.DataFrame(data["data"])
```

```python
df
```

```python
def setting(setting="functions"):
    base_url = f"http://r-function-orderbook:8203/api/functions/setting/{setting}"
    response = requests.get(base_url)

    if response.status_code == 200:
        return response.json()
    else:
        print(f"Error: {response.status_code}")
        return None
```

```python
# functions
f=setting("functions")
df = pd.DataFrame(f["data"])
df
```

```python
# catalogs
f=setting("catalogs")
df = pd.DataFrame(f["data"])
df
```

```python
# consumers
f=setting("consumers")
df = pd.DataFrame(f["data"])
df
```
